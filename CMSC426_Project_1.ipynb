1. Write Python code to cluster the orange ball using [Single Gaussian](https://cmsc426.github.io/colorseg/#gaussian) [30 points]

```python
# TODO: Import all python packages you need
import glob
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import gdown
import os
from skimage.measure import label, regionprops

# TODO: Read in training images

train_paths = glob.glob('/content/train_images/*.jpg') # find all image files and strore their paths in list
train_images = [mpimg.imread(p) for p in train_paths] # read each path in train_paths and convert each .jpg into a 3D array (640 rows, 480 cols, 3RBG vals), store in list

# TODO: Iterate over training images to extract orange pixels using masks

def circle_mask(shape, cx, cy, r): #function to create a circle mask
    rows, cols = shape[:2]
    yy, xx = np.mgrid[:rows, :cols]          # y and x coordinate of every pixel
    return (xx - cx)**2 + (yy - cy)**2 <= r**2

def show_with_grid(i): # function to show image with a grid
    plt.figure(figsize=(8, 10))
    plt.imshow(train_images[i])
    plt.title(train_paths[i].split('/')[-1])
    plt.xticks(range(0, 480, 20), fontsize=6)
    plt.yticks(range(0, 640, 20), fontsize=6)
    plt.grid(True, alpha=0.3)
    plt.show()

def show_mask(i): # function to show mask on grid
    name = train_paths[i].split('/')[-1]
    cx, cy, r = circles[name]
    img = train_images[i]
    mask = circle_mask(img.shape, cx, cy, r)

    plt.figure(figsize=(10, 6))
    plt.subplot(1, 2, 1)
    plt.imshow(img)
    plt.imshow(mask, alpha=0.4, cmap='gray')   # mask laid over the image
    plt.title(name)

    plt.subplot(1, 2, 2)
    plt.imshow(img * mask[:, :, None])  # only the pixels kept
    plt.xlim(cx - 40, cx + 40)    # zoom in on the ball
    plt.ylim(cy + 40, cy - 40)
    plt.title('Pixels kept')
    plt.show()

circles = { # dic with location and radius of circle for each image (manually went through each image)
    '68.jpg': (230, 350, 25),
    '76.jpg': (228, 330, 23),
    '91.jpg': (219, 297, 19),
    '99.jpg': (223, 285, 18),
    '106.jpg': (226, 275, 16),
    '114.jpg': (220, 269, 15),
    '121.jpg': (226, 264, 14),
    '137.jpg': (226, 253, 13),
    '144.jpg': (222, 245, 13),
    '152.jpg': (222, 241, 12),
    '160.jpg': (228, 238, 11),
    '168.jpg': (223, 235, 11),
    '176.jpg': (226, 230, 10),
    '192.jpg': (228, 227, 10),
    '200.jpg': (229, 223, 9),
    '208.jpg': (233, 223, 9),
    '216.jpg': (232, 221, 8),
    '223.jpg': (231, 219, 8),
    '231.jpg': (229, 214, 8),
    '248.jpg': (233, 211, 8),
    '256.jpg': (235, 209, 7),
    '264.jpg': (234, 208, 7),
    '280.jpg': (231, 205, 7)
}

i = train_paths.index('/content/train_images/280.jpg')
show_with_grid(i)
show_mask(i)

orange_pixels = [] # List to hold each image's ball pixels (orange)

for path, img in zip(train_paths, train_images):  # Each image adds one K×3 (K represents the number of pixels in the circle mask) array to orange_pixels, and every row in it is one orange pixel's [R, G, B] values
    name = path.split('/')[-1]
    cx, cy, r = circles[name]
    mask = circle_mask(img.shape, cx, cy, r)  #2d array where 1 is oragne and 0 is not
    orange_pixels.append(img[mask]) # this image's ball pixels, K x 3

orange_pixels = np.vstack(orange_pixels).astype(np.float64)   # stack all Kx3 arrays so we have an array of orange pixels from every image

# TODO: Compute mean and covariance using MLE(Maximum Likelihood Estimation)

N = orange_pixels.shape[0]  #number of orange pixels

mu = np.mean(orange_pixels, axis=0) # average R, G, B  -> shape (3,) (THE AVERAGE ORANGE)

diff = orange_pixels - mu # each pixel minus the mean -> N x 3 (HOW FAR EACH PIXEL IS FROM MU)
Sigma = (diff.T @ diff) / N # covariance -> 3 x 3 (AVG PRODUCT OF DEVIATIONS)

print("mu =", mu)
print("Sigma =\n", Sigma)

# TODO: Compute PDF(Probability Density Function) of single gaussian model

def gaussian_pdf(X, mu, Sigma):
    # X: N x 3 array of pixels, mu: (3,), Sigma: 3 x 3
    Sigma_inv = np.linalg.inv(Sigma)
    Sigma_det = np.linalg.det(Sigma)
    norm = 1.0 / np.sqrt((2 * np.pi)**3 * Sigma_det)

    diff = X - mu # N x 3
    exponent = -0.5 * np.sum((diff @ Sigma_inv) * diff, axis=1)   # N values
    return norm * np.exp(exponent)


# TODO: Set parameters (threshold, prior)

prior = 0.5 # Set from 426page link above

train_scores = gaussian_pdf(orange_pixels, mu, Sigma) * prior
tau = np.percentile(train_scores, 5)     # keep ~95% of the training orange pixels

# TODO: Send test images into algorithm to detect orange ball

test_paths = glob.glob('/content/test_images/*.jpg')
test_images = [mpimg.imread(p) for p in test_paths]

test_masks = []
for img in test_images:
    H, W, _ = img.shape
    pixels = img.reshape(-1, 3).astype(np.float64)   # (H*W) x 3, one row per pixel
    scores = gaussian_pdf(pixels, mu, Sigma) * prior  # likelihood x prior for each pixel
    mask = (scores >= tau).reshape(H, W)              # True where the pixel looks orange
    test_masks.append(mask)
```

2. Write Python code to cluster the orange ball using [Gaussian Mixture Model](https://cmsc426.github.io/colorseg/#gmm) [40 points] and estimate the [distance](https://cmsc426.github.io/colorseg/#distest) to the ball [20 points]. Also, plot all the GMM ellipsoids [10 points].


You are NOT allowed to use any built-in Python package(s) like *sklearn.mixture.GaussianMixture* for GMM. To help you with code implementation, we have given the pseudocode :-)

```python
def trainGMM(K):
  X = orange_pixels # array containing all orange pixels
  N = X.shape[0] # number of orange pixels
  eps = 0.1
  max_iters = 100

# TODO: Set convergence criteria and initialize scaling factor, gaussian mean and covariance
  scalars = np.ones(K) / K  # list of K ones and divide each 1 by K so that every element can sum to 1
  idx = np.random.choice(N, K, replace=False)
  means = X[idx]
  covariances = np.array([Sigma] * K) #take problem 1's Sigma K times

# TODO: Main training algorithm (EM algorithm)
  for it in range(max_iters):
    prev_means = means.copy() #make seperate copy of means array before m step

    # E-step (calculate how much each pixel "fits" with each gaussian)
    alpha = np.zeros((N, K))
    for i in range(K):
      alpha[:, i] = scalars[i] * gaussian_pdf(X, means[i], covariances[i])
    alpha = alpha / alpha.sum(axis=1, keepdims=True) #divide each row by total so add up to 1

    # M-step (update each gaussian using shares from E step - problem 1's mean and covariance, done once per gaussian with weighted pixels.)
    for i in range(K):
      w = alpha[:, i]
      scalars[i] = w.mean()
      means[i] = w @ X / w.sum()
      diff = X - means[i]
      covariances[i] = (w * diff.T) @ diff / w.sum()

    if np.abs(means - prev_means).sum() < eps:
      break

  return scalars, means, covariances

def testGMM(Model_parameters,threshold,prior):
  #load the model
  scalars = Model_parameters[0]
  means = Model_parameters[1]
  covariances = Model_parameters[2]

# TODO:  Read test images
  test_paths = glob.glob('/content/test_images/*.jpg')
  test_images = [mpimg.imread(p) for p in test_paths]

# TODO:  Main testing loop over all test images and use thresholding to get the orange ball
  cluster_parameters = []
  for img in test_images:
    pixels = img.reshape(-1, 3).astype(np.float64)
    scores = 0
    for i in range(len(scalars)):
      scores = scores + scalars[i] * gaussian_pdf(pixels, means[i], covariances[i])
    mask = (scores * prior >= threshold).reshape(640, 480)
    cluster_parameters.append(mask)

# TODO: Saving predictions to the result folder
  os.makedirs('results', exist_ok=True)
  for i in range(len(cluster_parameters)):
    plt.imsave('results/' + str(i) + '.png', cluster_parameters[i], cmap='gray')

  return cluster_parameters

def measureDepth(cluster_parameters):
  # hint: fit a simple parametric model (choose a model of your choice) to estimate distance from different parameters based on the image. For eg. area of the ball on the image decreases with distance (generally follows a inverse square curve).
# TODO: Identify the pixels which belong to the orange ball
  ball_areas = []
  for mask in cluster_parameters:
    regions = regionprops(label(mask))
    ball_areas.append(max(r.area for r in regions))   # area of the biggest blob = the ball
  ball_areas = np.array(ball_areas)

# TODO: Use the data provided and any feature you like to obtain a model to estimate distance
  train_distances = np.array([int(name[:-4]) for name in circles])
  train_areas = np.array([np.pi * r**2 for (cx, cy, r) in circles.values()])
  a = np.mean(train_areas * train_distances**2)

  distance = np.sqrt(a / ball_areas)
  return distance

def plotGMM(Model_parameters, distance):
  scalars, means, covariances = Model_parameters

# TODO: Visualize GMM ellipsoids
  ax = plt.figure().add_subplot(projection='3d')

  # points on a sphere of radius 1
  u = np.linspace(0, 2 * np.pi, 30)
  v = np.linspace(0, np.pi, 30)
  x = np.outer(np.cos(u), np.sin(v))
  y = np.outer(np.sin(u), np.sin(v))
  z = np.outer(np.ones(30), np.cos(v))
  sphere = np.array([x.flatten(), y.flatten(), z.flatten()])    # 3 x 900

  for i in range(len(scalars)):
    vals, vecs = np.linalg.eigh(covariances[i])
    pts = vecs @ np.diag(2 * np.sqrt(vals)) @ sphere + means[i].reshape(3, 1)
    ax.plot_surface(pts[0].reshape(30, 30), pts[1].reshape(30, 30), pts[2].reshape(30, 30), alpha=0.3)

  ax.set_xlabel('R')
  ax.set_ylabel('G')
  ax.set_zlabel('B')
  plt.show()
```

```python
# Main function (Algorithm 1 in the pseudocode above)


# -Write Python code to cluster the orange ball using Gaussian Mixture Model [40 points].


# -Estimate the distance to the ball [20 points].


# -Plot all the GMM ellipsoids [10 points].





# TODO: Import all python packages you need


K = 3             # number of gaussians
prior = 0.5
mode_flag = 1    # 0 = train, 1 = test

if mode_flag == 0:
  # training images and orange_pixels come from the Problem 1 cells
  scalars, means, covariances = trainGMM(K)

  # threshold: keep ~95% of the training orange pixels
  scores = 0
  for i in range(K):
    scores = scores + scalars[i] * gaussian_pdf(orange_pixels, means[i], covariances[i])
  threshold = np.percentile(scores * prior, 5)

  np.savez('gmm_model.npz', scalars=scalars, means=means, covariances=covariances, threshold=threshold)
  print("Training done")

else:
  model = np.load('gmm_model.npz')
  Model = (model['scalars'], model['means'], model['covariances'])
  threshold = model['threshold']

  cluster_parameters = testGMM(Model, threshold, prior)
  distance = measureDepth(cluster_parameters)
  print("Estimated distances (cm):", distance)
  plotGMM(Model, distance)

for img, mask in zip(test_images, cluster_parameters):
  plt.subplot(1, 2, 1)
  plt.imshow(img)
  plt.subplot(1, 2, 2)
  plt.imshow(mask, cmap='gray')
  plt.show()
```
