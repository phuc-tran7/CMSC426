### **Problem 2.1 (2 pts)**

What properties do projective transformations preserve?

- A) Collinearity and cross-ratio.
- B) Angles and distances.
- C) Areas and angles.
- D) Parallelism and distance.


Answer:

**Answer: A) Collinearity and cross-ratio.**

A projective transformation (homography) $x' = Hx$ maps lines to lines, so collinearity and incidence are preserved. The cross-ratio of four collinear points is its fundamental invariant:

$$
\text{Cross}(x_1, x_2, x_3, x_4) = \frac{|x_1 x_2|\,|x_3 x_4|}{|x_1 x_3|\,|x_2 x_4|}
$$

### **Problem 2.2 (6 pts)**

Show that an affine transformation can map a circle to elliipse, but cannot map an ellipse to a hyperbola or parabola.

Answer:
An affine transformation is $x' = Ax + t$.

**Circle → ellipse:** Stretch the unit circle $x^2 + y^2 = 1$ with $x' = 2x$, $y' = y$. This gives

$$
\frac{x'^2}{4} + y'^2 = 1,
$$

which is an ellipse.

**Ellipse → hyperbola or parabola:** An ellipse is bounded (it stays in a finite region). An affine map keeps finite points finite, so the image of an ellipse is still bounded. Hyperbolas and parabolas go off to infinity, so an affine map cannot turn an ellipse into either one.

## **Section 3 : Pinhole Camera Model** (10 pts)

### **Problem 3.1 (3 pts)**

 What is intrinsic matrix K? For each quantity of the intrinsic matrix

—state: (a) what it represents, (b) what units it uses

Answer:

The intrinsic matrix $K$ describes the inside of the camera. It converts a 3-D point seen by the camera into a pixel location in the image.

$$
K=\begin{bmatrix} f_x & s & c_x\\ 0 & f_y & c_y\\ 0 & 0 & 1 \end{bmatrix}
$$

- **$f_x, f_y$ (focal length):** (a) How strongly the camera zooms in the $x$ and $y$ directions. (b) Pixels.
- **$s$ (skew):** (a) How slanted the pixel grid is. It is usually $0$. (b) Pixels.
- **$c_x, c_y$ (principal point):** (a) The pixel where the center of the lens points, usually near the middle of the image. (b) Pixels.
- **$1$:** (a) A constant that makes the matrix work with homogeneous coordinates. (b) No units.

### **Problem 3.2 (2 pts)**
he intrinsic camera matrix is known for images of size H × W . If the
image is resized to H/
2 × W/
2 (half the original), do we need to scale the camera matrix
by 0.5 in order to keep the same relation?

Answer:

**No.** Multiplying all of $K$ by $0.5$ does nothing, because in homogeneous coordinates $0.5K$ gives the same pixels as $K$.

When the image is halved, pixel coordinates are halved too. So we halve every entry except the last row:

$$
K' = \begin{bmatrix} f_x/2 & s/2 & c_x/2\\ 0 & f_y/2 & c_y/2\\ 0 & 0 & 1 \end{bmatrix}
$$

### **Problem 3.4 (2 pts)**
**(a)** Define a *vanishing point*.

**(b)** How is the vanishing point projected onto the image plane? Give the expression in homogeneous coordinates.

Answer:

**(a)** A vanishing point is the point in the image where parallel lines in 3-D appear to meet. It is the image of the point at infinity in the direction of those lines.

**(b)** A direction $d = (d_1, d_2, d_3)^T$ in 3-D corresponds to the point at infinity $D = (d_1, d_2, d_3, 0)^T$. Projecting it with $P = K[R \mid t]$:

$$
v = P \begin{bmatrix} d\\ 0 \end{bmatrix} = K R\, d
$$

The translation $t$ drops out, so the vanishing point depends only on the direction of the lines, not their position.

The optical flow of an image can be decomposed into two components: translational
flow, which arises due to the linear motion of the camera, and rotational flow, which is a
result of the camera’s angular motion. Mention each quantity stands for what.
$$
\dot{\mathbf p}
= \frac{1}{Z}
\begin{bmatrix}
xV_z - V_x \\
yV_z - V_y
\end{bmatrix}
\;+\;
\begin{bmatrix}
xy & -(1+x^2) & y \\
(1+y^2) & -xy & -x
\end{bmatrix}
\begin{bmatrix}
\Omega_x\\
\Omega_y\\
\Omega_z
\end{bmatrix}
$$

Answer:


- **$\dot{\mathbf{p}}$:** The optical flow, i.e. the velocity $(\dot{x}, \dot{y})$ of the image point $\mathbf{p}$.
- **$x, y$:** Coordinates of the image point $\mathbf{p}$ in normalized image coordinates (focal length $= 1$).
- **$Z$:** Depth of the 3-D point, i.e. its distance from the camera along the optical axis.
- **$V_x, V_y, V_z$:** The camera's linear (translational) velocity along the $x$, $y$, and $z$ axes.
- **$\Omega_x, \Omega_y, \Omega_z$:** The camera's angular (rotational) velocity about the $x$, $y$, and $z$ axes.

The first term is the **translational flow**, which shrinks as depth $Z$ grows (far points move less). The second term is the **rotational flow**, which does not depend on depth.

### **Problem 5.1 (4 pts)**

Consider a homography transformation that maps a point \(P\) in the 2D plane to a new point \(P'\) using the homography matrix \(H\). The transformation is:

$$
P' = H \, P
$$

where $P = (x, y, 1)^{\mathsf T}$ is a point in homogeneous coordinates, and \(H\) is a $3 \times 3$ homography matrix.

Suppose we apply the following homography matrix \(H\) to a point \(P\):

$$
H =
\begin{bmatrix}
2 & 0 & 3 \\
1 & 1 & -4 \\
0 & 0 & 1
\end{bmatrix}
$$

Given that

$$
P = (1, 2, 1)^{\mathsf T},
$$

calculate the transformed point \(P'\) after applying the homography. Express the result in **Cartesian coordinates**.


Answer:


$$
P' = HP = \begin{bmatrix} 2 & 0 & 3\\ 1 & 1 & -4\\ 0 & 0 & 1 \end{bmatrix}
\begin{bmatrix} 1\\ 2\\ 1 \end{bmatrix}
= \begin{bmatrix} 2(1) + 0(2) + 3(1)\\ 1(1) + 1(2) - 4(1)\\ 0(1) + 0(2) + 1(1) \end{bmatrix}
= \begin{bmatrix} 5\\ -1\\ 1 \end{bmatrix}
$$

To get Cartesian coordinates, divide by the last entry, which is $1$:

$$
P' = (5,\ -1)
$$

### **Problem 5.2 — Vanishing Line under a Homography (6 pts)**

You have a flat **world plane** (e.g., a tabletop). Points on this plane are written in homogeneous coordinates as
$$
X = (X, Y, W)^\top .
$$

A camera maps points on the plane to the image via a $3\times 3$ homography \(H\):
$$
x' \sim H\,X ,
$$
where $\sim$ denotes equality up to scale.

The **line at infinity** on the world plane is
$$
\ell_\infty = (0, 0, 1)^\top .
$$
Its image is called the **vanishing line** (the “horizon”) and we denote it by $\ell'_\infty$.

**Task.**  
Find the vanishing line in the image. Show that
$$
\ell'_\infty \;\sim\; H^{-\top}\,\ell_\infty .
$$





Answer:

A point $X$ lies on a line $\ell$ when $\ell^T X = 0$.

Insert $H^{-1}H = I$:

$$
\ell^T H^{-1} H X = 0
\quad\Rightarrow\quad
(H^{-T}\ell)^T (HX) = 0
\quad\Rightarrow\quad
(H^{-T}\ell)^T x' = 0
$$

So every image point $x' = HX$ lies on the line $H^{-T}\ell$. That means lines transform as

$$
\ell' \sim H^{-T}\ell
$$

Applying this to the line at infinity $\ell_\infty = (0, 0, 1)^T$:

$$
\ell'_\infty \sim H^{-T}\ell_\infty = H^{-T}\begin{bmatrix} 0\\ 0\\ 1 \end{bmatrix}
$$

That is, the vanishing line is the third column of $H^{-T}$, which is the third row of $H^{-1}$.
